```sql
SELECT *
FROM `retail`.`sales`.`retail_sales_dataset`;

--Validating answers from the Retails Analysis Bot
--- Which gender spent the most on average?

SELECT Gender, AVG(`Total Amount`) AS avg_spend
FROM `retail`.`sales`.`retail_sales_dataset`
GROUP BY Gender; 
-- The response that Retails Analysis Bot gave was correct.
-- This is why:

-- Columns you want to see is the average amount spent, and the gender (male and female).
-- In the SELECT statement, you will aggregate the `Total Amount`(numerical column), 
    and also write the unique column (Gender).
    You GROUP BY the unique column, 'Gender'. 



-- Which customer age group contributed the most revenue.

-- Group the age groups as follows:
        18-24- Young adults
        25-34- Early career adults
        35-44- Mid-Age adults
        45-54 - Mature adults
        55-64- Senior Adults

SELECT 
    CASE WHEN Age BETWEEN 18 AND 24 THEN 'Young adults'
         WHEN Age BETWEEN 25 AND 34 THEN 'Early career adults'
         WHEN Age BETWEEN 35 AND 44 THEN 'Mid-Age adults'
         WHEN Age BETWEEN 45 AND 54 THEN 'Mature adults'
         WHEN Age BETWEEN 55 AND 64 THEN 'Senior adults'
    END AS age_group,
    SUM(`Total Amount`) AS total_revenue

FROM `retail`.`sales`.`retail_sales_dataset`
GROUP BY 
        CASE WHEN Age BETWEEN 18 AND 24 THEN 'Young adults'
         WHEN Age BETWEEN 25 AND 34 THEN 'Early career adults'
         WHEN Age BETWEEN 35 AND 44 THEN 'Mid-Age adults'
         WHEN Age BETWEEN 45 AND 54 THEN 'Mature adults'
         WHEN Age BETWEEN 55 AND 64 THEN 'Senior adults'
    END
ORDER BY total_revenue DESC;

-- What the Retails Analysis Bot answered was correct.
-- First, I used the CASE statement to find the categorical column for the Age group.
-- Then Aggregate the `Total Amount` to find the SUM
-- I needed needed to GROUP BY the CASE statement for the Age group again.
-- Finally, ORDER BY the aggregated column.



-- Total revenue spent per product category

SELECT SUM(`Total Amount`) AS total_amount_spent,
         `Product Category`
FROM `retail`.`sales`.`retail_sales_dataset`
GROUP BY `Product Category`;

-- This is why Retail Analysis Bot is correct:
-- I had to aggregate on the SELECT statement the `Total Amount` and used the alias total_amount_spent.
-- Use the unique column which is the `Product Category`.
-- Lastly, GROUP BY the unique column.


```
